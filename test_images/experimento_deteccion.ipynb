```python
import numpy as np
from plantcv import plantcv as pcv
from plantcv.parallel import WorkflowInputs
```

```python
import cv2

# Load your image
image_path = "./gemini_puntos.jpg"
img = cv2.imread(image_path)

print("Click on the image to get RGB values. Press 'q' or the ESC key to close the window.")
print("-" * 50)

def get_pixel_color(event, x, y, flags, param):
    if event == cv2.EVENT_LBUTTONDOWN:
        # OpenCV loads images in BGR format, so we reverse it to RGB
        b, g, r = img[y, x]
        print(f"{r},{g},{b}")

cv2.namedWindow("Click to Pick Colors")
cv2.setMouseCallback("Click to Pick Colors", get_pixel_color)

while True:
    cv2.imshow("Click to Pick Colors", img)
    # Wait for a key press
    key = cv2.waitKey(1) & 0xFF
    if key == 27 or key == ord('q'): 
        break

cv2.destroyAllWindows()
```

```python
# Input/Output options

# plantcv-utils inspect_pixels --image ./gemini_puntos.jpg --outfile pixel_samples.txt

args = WorkflowInputs(
    #images=["img/color_image.jpg"],    
    images=["./gemini_puntos.jpg"],   
    names="image1",
    result="ml_tutorial_results.json",
    outdir=".",
    writeimg=True,
    debug="plot"
    )

# Set debug as a global parameter 
pcv.params.debug = args.debug
```

```python
img, path, filename = pcv.readimage(filename=args.image1)
```

```python
mask = pcv.naive_bayes_classifier(rgb_img=img, 
                                  pdf_file="./machine_learning.txt")
```
